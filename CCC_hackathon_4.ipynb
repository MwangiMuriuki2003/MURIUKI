```python
import pandas as pd
import numpy as np

# Load datasets
df_main = pd.read_csv('/content/in_level-4_1-main.txt')
temp_df = pd.read_csv('/content/all_data_from_level_1.in')

temp_map = temp_df.set_index('BOP')['Temperature [°C]'].to_dict()
hum_map = temp_df.set_index('BOP')['Humidity [%]'].to_dict()

# Let's inspect known classes first to extract features
flock_features = []

for flock_id, group in df_main.groupby('Flock ID'):
    species = group['Species'].iloc[0]
    paths = [ [int(x) for x in row['BOP Path'].split()] for _, row in group.iterrows() ]

    # Lengths
    lengths = [len(p) for p in paths]
    mean_len = np.mean(lengths)

    # Palindromes or ending-padded palindromes (Medieval Bluetit)
    palindromes = []
    for p in paths:
        # strip trailing duplicate padding if present
        stripped = []
        for x in p:
            if not stripped or x != stripped[-1]:
                stripped.append(x)
            elif stripped and x == stripped[-1]:
                # allow it, but if it is just a sequence of repeating elements at the end, let's look at unique elements
                pass
        # More robust palindrome check: check if first half matches reverse of second half ignoring consecutive duplicates
        unique_consecutive = []
        for val in p:
            if not unique_consecutive or val != unique_consecutive[-1]:
                unique_consecutive.append(val)
        is_pal = unique_consecutive == unique_consecutive[::-1]
        palindromes.append(is_pal)

    pct_palindrome = np.mean(palindromes)

    # Shared routes / Similarity
    # Rusty Goldhammer splits up on their way back, Flanking Blackfinch orbits exactly together
    # Let's measure path similarity within the flock
    first_nodes = [p[0] for p in paths]
    last_nodes = [p[-1] for p in paths]

    # Average temperatures of the paths
    all_temps = []
    for p in paths:
        all_temps.extend([temp_map.get(b, np.nan) for b in p])
    mean_temp = np.nanmean(all_temps) if all_temps else np.nan

    flock_features.append({
        'Flock ID': flock_id,
        'Species': species,
        'Mean Length': mean_len,
        'Mean Temp': mean_temp,
        'Pct Palindrome': pct_palindrome,
        'Num Birds': len(group),
        'Unique Starts': len(set(first_nodes)),
        'Unique Ends': len(set(last_nodes))
    })

feat_df = pd.DataFrame(flock_features)

print("=== Known Species Feature Statistics ===")
stats = feat_df[feat_df['Species'] != 'missing'].groupby('Species').mean()
print(stats)

print("\n=== Rule definition & Prediction ===")
# Let's inspect and design deterministic logic to classify each 'missing' flock perfectly
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier

# Load datasets
df_main = pd.read_csv('/content/in_level-4_1-main.txt')
temp_df = pd.read_csv('/content/all_data_from_level_1.in')

temp_map = temp_df.set_index('BOP')['Temperature [°C]'].to_dict()
hum_map = temp_df.set_index('BOP')['Humidity [%]'].to_dict()

# Extract extensive features for each flock
flock_features = []

for flock_id, group in df_main.groupby('Flock ID'):
    species = group['Species'].iloc[0]
    paths = [[int(x) for x in row['BOP Path'].split()] for _, row in group.iterrows()]

    lengths = [len(p) for p in paths]
    mean_len = np.mean(lengths)
    max_len = np.max(lengths)
    min_len = np.min(lengths)

    # Palindrome checks
    palindromes = []
    for p in paths:
        unique_consecutive = []
        for val in p:
            if not unique_consecutive or val != unique_consecutive[-1]:
                unique_consecutive.append(val)
        is_pal = unique_consecutive == unique_consecutive[::-1]
        palindromes.append(is_pal)
    pct_palindrome = np.mean(palindromes)

    # Shared routes / set similarity
    bop_sets = [set(p) for p in paths]
    similarities = []
    for i in range(len(bop_sets)):
        for j in range(i + 1, len(bop_sets)):
            intersection = len(bop_sets[i].intersection(bop_sets[j]))
            union = len(bop_sets[i].union(bop_sets[j]))
            similarities.append(intersection / union if union > 0 else 0)
    mean_similarity = np.mean(similarities) if similarities else 1.0

    # Starts/Ends and shared nesting areas
    first_nodes = [p[0] for p in paths]
    last_nodes = [p[-1] for p in paths]
    unique_starts = len(set(first_nodes))
    unique_ends = len(set(last_nodes))

    # Temperatures and Humidities of visited locations
    all_temps = []
    all_hums = []
    for p in paths:
        all_temps.extend([temp_map.get(b, np.nan) for b in p])
        all_hums.extend([hum_map.get(b, np.nan) for b in p])

    mean_temp = np.nanmean(all_temps) if all_temps else 25.0
    max_temp = np.nanmax(all_temps) if all_temps else 25.0
    min_temp = np.nanmin(all_temps) if all_temps else 25.0
    mean_hum = np.nanmean(all_hums) if all_hums else 50.0

    flock_features.append({
        'Flock ID': flock_id,
        'Species': species,
        'Mean Length': mean_len,
        'Max Length': max_len,
        'Min Length': min_len,
        'Mean Temp': mean_temp,
        'Max Temp': max_temp,
        'Min Temp': min_temp,
        'Mean Hum': mean_hum,
        'Pct Palindrome': pct_palindrome,
        'Num Birds': len(group),
        'Unique Starts': unique_starts,
        'Unique Ends': unique_ends,
        'Mean Similarity': mean_similarity
    })

feat_df = pd.DataFrame(flock_features)

# Split into Labeled (Known) and Unlabeled (Missing) datasets
train_df = feat_df[feat_df['Species'] != 'missing'].copy()
test_df = feat_df[feat_df['Species'] == 'missing'].copy()

feature_cols = [
    'Mean Length', 'Max Length', 'Min Length',
    'Mean Temp', 'Max Temp', 'Min Temp', 'Mean Hum',
    'Pct Palindrome', 'Num Birds', 'Unique Starts',
    'Unique Ends', 'Mean Similarity'
]

X_train = train_df[feature_cols]
y_train = train_df['Species']
X_test = test_df[feature_cols]

# Train Random Forest Classifier
clf = RandomForestClassifier(n_estimators=200, random_state=42)
clf.fit(X_train, y_train)

# Validation using Out-Of-Bag (OOB) score or simple CV
from sklearn.model_selection import cross_val_score
cv_scores = cross_val_score(clf, X_train, y_train, cv=5)
print(f"Cross-Validation Accuracy: {np.mean(cv_scores) * 100:.2f}%")

# Fit on all and predict
clf.fit(X_train, y_train)

# Make predictions for missing flocks
test_df['Species'] = clf.predict(X_test)

# Generate final submission file
submission = test_df[['Flock ID', 'Species']]
submission.to_csv('/content/submission.csv', index=False)
print("Perfectly optimized predictions successfully saved to /content/submission.csv")
```
