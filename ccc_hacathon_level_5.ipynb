```python
import polars as pl
import lightgbm as lgb
import numpy as np
import gc

print("1. Loading Data with Strict Memory-Efficient Types...")
schema = {
    "Day": pl.Int32,
    "BOP": pl.Int32,
    "Arrivals": pl.Float32,
    "Occupancy": pl.Float32,
    "Wind X [m/s]": pl.Float32,
    "Wind Y [m/s]": pl.Float32,
    "Insects Delta [g/m²]": pl.Float32
}

df = pl.read_csv("in_level-5_1-main.txt", null_values="missing", schema_overrides=schema)
df = df.rename({
    "Wind X [m/s]": "Wind_X",
    "Wind Y [m/s]": "Wind_Y",
    "Insects Delta [g/m²]": "Insects_Delta"
})

print("2. Processing Level 4 Paths...")
paths_df = pl.read_csv("all_data_from_level_4.in", null_values="missing")

edges = []
path_freq = {}
for row in paths_df.iter_rows():
    bops = [int(x) for x in str(row[1]).split()]
    for i in range(1, len(bops)):
        edges.append({"Target_BOP": bops[i], "Source_BOP": bops[i-1]})
    for bop in bops:
        path_freq[bop] = path_freq.get(bop, 0) + 1

edges_df = pl.DataFrame(edges, schema={"Target_BOP": pl.Int32, "Source_BOP": pl.Int32})

# FIX: Drop self-loops so a BOP's own lag doesn't pollute its upstream stats
self_loops_count = len(edges_df.filter(pl.col("Source_BOP") == pl.col("Target_BOP")))
print(f"   Self-loops detected and removed: {self_loops_count}")
edges_df = edges_df.filter(pl.col("Source_BOP") != pl.col("Target_BOP"))

freq_df = pl.DataFrame(
    [{"BOP": k, "Path_Frequency": v} for k, v in path_freq.items()],
    schema={"BOP": pl.Int32, "Path_Frequency": pl.Int32}
)

del paths_df
gc.collect()

print("3. Engineering Core Features...")
df = df.sort(["BOP", "Day"])

df = df.with_columns([
    (pl.col("Day") % 365).cast(pl.Int32).alias("Day_of_Year"),
    pl.col("Occupancy").shift(1).over("BOP").cast(pl.Float32).alias("Lag1_Occupancy"),
    pl.col("Arrivals").shift(1).over("BOP").cast(pl.Float32).alias("Lag1_Arrivals"),
    pl.col("Arrivals").shift(7).over("BOP").cast(pl.Float32).alias("Lag7_Arrivals"),
]).fill_null(0)

train_df = df.filter(pl.col("Day") <= 730)
test_df = df.filter(pl.col("Day") > 730)

# Compute stable BOP historical baselines from training data ONLY
bop_stats = train_df.group_by("BOP").agg([
    pl.col("Arrivals").mean().cast(pl.Float32).alias("BOP_Mean_Arrivals"),
    pl.col("Occupancy").mean().cast(pl.Float32).alias("BOP_Mean_Occupancy"),
])

train_df = train_df.join(bop_stats, on="BOP", how="left")
test_df = test_df.join(bop_stats, on="BOP", how="left")

# FIX (the original bug): Join Path_Frequency to BOTH splits, not just test
train_df = train_df.join(freq_df, on="BOP", how="left")
test_df  = test_df.join(freq_df, on="BOP", how="left")

del df, bop_stats, freq_df
gc.collect()

print("4. Computing Upstream Features (Memory-Safe Join)...")
source_stats = train_df.select(["Day", "BOP", "Lag1_Occupancy", "Lag1_Arrivals"])

joined = edges_df.join(source_stats, left_on="Source_BOP", right_on="BOP", how="inner")

upstream = joined.group_by(["Day", "Target_BOP"]).agg([
    pl.col("Lag1_Occupancy").sum().cast(pl.Float32).alias("Upstream_Occ_Sum"),
    pl.col("Lag1_Occupancy").mean().cast(pl.Float32).alias("Upstream_Occ_Mean"),
    pl.col("Lag1_Arrivals").sum().cast(pl.Float32).alias("Upstream_Arr_Sum"),
])

del joined, source_stats, edges_df
gc.collect()

train_df = train_df.join(upstream, left_on=["Day", "BOP"], right_on=["Day", "Target_BOP"], how="left")
test_df = test_df.join(upstream, left_on=["Day", "BOP"], right_on=["Day", "Target_BOP"], how="left")

del upstream
gc.collect()

print("5. Finalizing Feature Sets & Sample Weighting...")
# FIX: Removed "Occupancy" from feature_cols (it's a label leak — used only for post-hoc blending)
feature_cols = [
    "Wind_X", "Wind_Y", "Insects_Delta",
    "Day_of_Year", "Path_Frequency",
    "Lag1_Occupancy", "Lag1_Arrivals", "Lag7_Arrivals",
    "Upstream_Occ_Sum", "Upstream_Occ_Mean", "Upstream_Arr_Sum",
    "BOP_Mean_Arrivals", "BOP_Mean_Occupancy"
    # NOTE: 'BOP' is intentionally REMOVED from features to prevent categorical overfitting
    # NOTE: 'Occupancy' is intentionally REMOVED to avoid label leak; used in step 7 blend only
]

# Sanity check: confirm all feature columns exist in both frames
missing_train = [c for c in feature_cols if c not in train_df.columns]
missing_test  = [c for c in feature_cols if c not in test_df.columns]
if missing_train or missing_test:
    raise RuntimeError(
        f"Missing feature columns!\n  train: {missing_train}\n  test: {missing_test}"
    )

train_df = train_df.select(feature_cols + ["Arrivals", "Day"]).fill_null(0)
test_df  = test_df.select(feature_cols + ["Day", "BOP", "Occupancy"]).fill_null(0)  # Keep BOP and Occupancy for final blending

X_train = train_df.select(feature_cols).to_pandas()
y_train = train_df.select("Arrivals").to_pandas()["Arrivals"].values
X_test  = test_df.select(feature_cols).to_pandas()

# FIX 1: Sample Weighting. Force the model to care 5x more about high-arrival BOPs.
threshold = np.quantile(y_train, 0.85)  # Top 15% of arrivals
sample_weights = np.where(y_train > threshold, 5.0, 1.0)

del train_df
gc.collect()

print("6. Training LightGBM Model (Weighted for Top-50 Ranking)...")
params = {
    'objective': 'regression',
    'metric': 'rmse',
    'boosting_type': 'gbdt',
    'num_leaves': 63,
    'learning_rate': 0.05,
    'feature_fraction': 0.8,
    'bagging_fraction': 0.8,
    'bagging_freq': 5,
    'verbose': -1,
    'n_estimators': 300,
    'random_state': 42
    # FIX 2: Removed 'categorical_feature': ['BOP'] to force generalization
}

model = lgb.LGBMRegressor(**params)
model.fit(X_train, y_train, sample_weight=sample_weights)

del X_train, y_train, sample_weights
gc.collect()

print("7. Predicting and Formatting Submission...")
predictions = model.predict(X_test)

# Add predictions back to test_df
test_df = test_df.with_columns(
    pl.Series("Predicted_Arrivals", predictions.astype(np.float32))
)

# FIX 3: Occupancy Blending.
# Since Occupancy has a 0.711 correlation with Arrivals, we boost the score
# of BOPs that already have high known occupancy on the test day.
test_df = test_df.with_columns(
    (pl.col("Predicted_Arrivals") + (pl.col("Occupancy") * 0.2)).cast(pl.Float32).alias("Final_Score")
)

submission = (
    test_df.sort(by=["Day", "Final_Score"], descending=[False, True])
    .group_by("Day")
    .head(50)
    .group_by("Day")
    .agg(
        pl.col("BOP").cast(pl.String).str.join(" ").alias("Top 50 Arrivals BOPs")
    )
    .sort("Day")
)

# Convert to Pandas for exact header formatting (no quotes)
submission_pd = submission.to_pandas()
submission_pd.columns = ["Day", "Top 50 Arrivals BOPs"]
submission_pd.to_csv("submission.csv", index=False)

print("✅ Success! 'submission.csv' has been generated and is ready for upload.")
print("\nPreview of submission.csv:")
print(submission_pd.head(3))
```
