```python
import os

# Recursively find any ZIP or data files containing level-5 or level_5
found_files = []
for root, dirs, files in os.walk('/'):
    for file in files:
        if 'level-5' in file or 'level_5' in file or file.endswith('.zip'):
            if not root.startswith('/usr') and not root.startswith('/lib') and not root.startswith('/sys') and not root.startswith('/proc'):
                found_files.append(os.path.join(root, file))

print("Matching files found:")
for f in found_files:
    print(f)
```
